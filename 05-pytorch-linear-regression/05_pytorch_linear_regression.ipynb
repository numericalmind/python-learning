## 1. Import Libraries

PyTorch is used to build and train the linear regression model.


```python
import torch
import torch.nn as nn
```

```python
torch.manual_seed(42);
```

## 2. Create the Dataset

The dataset follows the linear relationship:

$$
y = 2x + 1
$$

The goal is to train the model without directly providing the values 2 and 1.


```python
x = torch.tensor([
    [1.0],
    [2.0],
    [3.0],
    [4.0]
])

y = torch.tensor([
    [3.0],
    [5.0],
    [7.0],
    [9.0]
])
```

## 3. Build the Model

A simple linear model is used:

$$
\hat{y} = wx + b
$$

The model will learn the weight `w` and bias `b` during training.


```python
model = nn.Linear(1,1)
```

## 4. Define the Loss Function and Optimizer

Mean Squared Error (MSE) measures the difference between predictions
and actual values.

Stochastic Gradient Descent (SGD) updates the model parameters
using the calculated gradients.


```python
loss_function = nn.MSELoss()
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)
```

## 5. Train the Model

Each training iteration consists of:

1. Forward pass
2. Loss calculation
3. Clearing previous gradients
4. Backpropagation
5. Parameter update


```python
print("Initial weight: " , model.weight.item())
print("Initial bias:", model.bias.item())
```

```python
epochs = 1500

for epoch in range(epochs):
  predictions = model(x)
  loss = loss_function(predictions, y)

  optimizer.zero_grad()
  loss.backward()
  optimizer.step()

  if epoch % 300 == 0:
    print(f"Epoch: {epoch}, Loss: {loss.item():.6f}")
    print("Weight:", model.weight.item())
    print("Bias:", model.bias.item())
```

## 6. Evaluate the Results

After training, the learned parameters are compared with the
true relationship:

$$
y = 2x + 1
$$

Therefore, the expected values are approximately:

- Weight = 2
- Bias = 1

```python
print("Final loss:", loss.item())
print("Learned weight:", model.weight.item())
print("Learned bias:", model.bias.item())

with torch.no_grad():
    final_predictions = model(x)

print("\nPredictions:")
print(final_predictions)

print("\nActual values:")
print(y)
```
